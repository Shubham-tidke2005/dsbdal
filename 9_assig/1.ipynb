1. Load the demo dataset in dataframe object df
2. Detect the outlier using BoxPlot.
3. Handle the outlier using Quantile based flooring and capping (Hint: the outlier is capped at a certain value above the 90th percentile value
or floored at a factor below the 10th percentile value)

```python
import pandas as pd     
import numpy as np   
import seaborn as sns   
import matplotlib.pyplot as plt           
```

```python
df=pd.read_csv("demo1.csv")
df.head(5)
```

```python
#plt.boxplot(df["reading score"])
plt.figure(figsize=(10,5))
sns.boxplot(data=df)
plt.show()
```

```python
sns.boxplot(data=df["reading score"])
plt.show()
```

```python
# Step 3: Quantiles
lower = df["reading score"].quantile(0.10)
upper = df["reading score"].quantile(0.90)
```

```python
# Step 4: Flooring & Capping
df["reading score"] = np.where(df["reading score"] < lower, lower, df["reading score"])
df["reading score"] = np.where(df["reading score"] > upper, upper, df["reading score"])
```

```python
sns.boxplot(data=df["reading score"])
plt.show()
```
