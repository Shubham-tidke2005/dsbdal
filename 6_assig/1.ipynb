# 1. Creation of Dataset " StudentsPerformance" using Microsoft Excel.
The features of dataset are: Math_Score, Reading_Score, Writing_Score, Placement_Score, Club_Join_Date
Range of Values: Math_Score [60-80], Reading_Score[75-,95], ,Writing_Score [60,80], Placement_Score[75-100], Club_Join_Date [2018-2021].
Response VAriable : Placement Score
In 20% data, fill the impurities.

```python
import pandas as pd     
import numpy as np   
```

```python
np.random.seed(0)
n=100
```

```python
data_dic={
    "Math_Score": np.random.randint(60,80,n) ,
    "Reading_Score": np.random.randint(60,80,n) ,
    "Writing_Score":  np.random.randint(60,80,n) ,
    "Placement_Score": np.random.randint(60,80,n) ,
    "Club_Join_Date": np.random.choice(pd.date_range("2018-1-1","2021-12-30"),n)
}
```

```python
df=pd.DataFrame(data_dic)
df.head(5)
```

```python
for col in df.columns:
     df.loc[np.random.choice(df.index, int(0.2*n), replace=False), col] = np.nan
```
