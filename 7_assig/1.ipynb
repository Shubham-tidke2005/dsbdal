1. Load the Academic Performance dataset in data frame object.
2. Check null values in the dataset.
3. Check missing values in dataset and replace the null values with standard null value NaN
4. Replace the missing value of Math Score with Mean Value
5. Replace the missing value of Reading Score with standard deviation
6. Replace the missing value of place with common value "Nashik"

```python
import pandas as pd     
import numpy as np   
```

```python
df=pd.read_csv("StudentsPerformanceTest1.csv")
df.head(5)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.replace(["na","NaN","null","Na"],np.nan,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
ndf=df.dropna(subset=["math score"])
ndf["math score"]=ndf["math score"].astype(int)
ms_mean=int(ndf["math score"].mean())
df["math score"]=df["math score"].replace(np.nan,str(ms_mean))
df["math score"]=df["math score"].astype(int)

df
```

```python
ndf=df.dropna(subset=["reading score"])
ndf["reading score"]=ndf["reading score"].astype(int)
ms_mean=int(ndf["reading score"].std())
df["reading score"]=df["reading score"].replace(np.nan,str(ms_mean))
df["reading score"]=df["reading score"].astype(int)
df
```

```python
df["Region"]=df["Region"].fillna("Nashik",inplace=True)
df
```
