```python
import pandas as pd   
import numpy as np          
import seaborn as sns      
import matplotlib.pyplot as plt   
```

```python
data = {
    "gender": ["Male", "Female", "Female", "Male", None, "Female", "Male", None],
    "score": [85, 90, None, 78, 88, None, 92, 75],
    "age": [18, 19, 18, None, 20, 19, None, 18]
}
```

```python
df=pd.DataFrame(data)
df.head(5)
```

# 1. Check missing values in whole dataframe

```python
df.isnull()
```

 # 2. Missing values column-wise

```python
df.isnull().sum()
```

# 3. count row wise missing value

```python
df.isnull().sum(axis=1)
```

# Missing values in gender column

```python
df["gender"].isnull().sum()
```

# Groupby count of missing values (gender & score)

```python
df.groupby("gender")["score"].apply(lambda x:x.isnull().sum())
```

# Replace missing values in score column with mean

```python
df["score"]=df["score"].fillna(df["score"].mean(),inplace=True)
df["age"]=df["age"].interpolate()
df["gender"]=df["gender"].ffill()
```
