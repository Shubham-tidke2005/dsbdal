1. Load the demo dataset in dataframe object df
2. Detect the outlier using Inter Quantile Range(IQR)
3. remove the outliers from the dataset.

```python
import pandas as pd     
import numpy as np   
import seaborn as sns   
import matplotlib.pyplot as plt           
```

```python
df=pd.read_csv("demo1.csv")
df.head(5)
```

```python
sns.boxplot(data=df["placement score"])
```

```python
Q1 = df["placement score"].quantile(0.25)
Q3 = df["placement score"].quantile(0.75)
IQR = Q3 - Q1

# Step 3: Calculate bounds
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
df_new = df[(df["placement score"] >= lower) & (df["placement score"] <= upper)]
df_new.head(5)
```

```python
sns.boxplot(data=df_new["placement score"])
```
