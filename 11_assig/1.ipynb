1. Load the demo dataset in dataframe object df
2. Detect the outlier using Z-score
3. replace the outliers with the median value.


```python
import pandas as pd     
import numpy as np   
import seaborn as sns   
import matplotlib.pyplot as plt           
```

```python
df=pd.read_csv("demo1.csv")
df.head(5)
```

```python
sns.scatterplot(data=df["placement score"])
plt.show()
```

```python
plt.scatter(df["placement score"],df["placement score"])
plt.show()
```

```python
plt.scatter(df["math score"], df["placement score"])
plt.xlabel("Math Score")
plt.ylabel("Placement Score")
plt.title("Scatter Plot for Outlier Detection")
plt.show()
```

```python
lower = df["placement score"].quantile(0.10)
upper = df["placement score"].quantile(0.90)
```

```python
df["placement score"] = np.where(df["placement score"] < lower, lower, df["placement score"])
df["placement score"] = np.where(df["placement score"] > upper, upper, df["placement score"])
```

```python
plt.scatter(df["placement score"],df["placement score"])
plt.show()
```
