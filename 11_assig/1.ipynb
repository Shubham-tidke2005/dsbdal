1. Load the demo dataset in dataframe object df
2. Detect the outlier using Z-score
3. replace the outliers with the median value.


```python
import pandas as pd     
import numpy as np   
import seaborn as sns   
import matplotlib.pyplot as plt           
```

```python
df=pd.read_csv("demo1.csv")
df.head(5)
```

```python
for i in df.columns:
    sns.boxplot(data=df[i])
    plt.show()
```

```python
for col in df.columns:
    mean = df[col].mean()
    std = df[col].std()
    
    z = (df[col] - mean) / std
    
    # Step 3: Replace outliers with median
    median = df[col].median()
    df[col] = np.where(abs(z) > 3, median, df[col])

print("\nUpdated Data (Outliers Replaced):\n", df)
```
