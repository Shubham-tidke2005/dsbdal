```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from sklearn.feature_extraction.text import TfidfVectorizer

# Download packages
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# Sample text
text = "Python is used for Natural Language Processing"

print("Original Text:")
print(text)
```

```python
# 1. Tokenization
tokens = word_tokenize(text)
print("\nTokens:")
print(tokens)

# 2. POS Tagging
print("\nPOS Tags:")
print(pos_tag(tokens))

# 3. Stop Words Removal
stop_words = stopwords.words('english')

filtered = [word for word in tokens if word.lower() not in stop_words]

print("\nAfter Stop Words Removal:")
print(filtered)

# 4. Stemming
ps = PorterStemmer()

stemmed = [ps.stem(word) for word in filtered]

print("\nStemmed Words:")
print(stemmed)

# 5. Lemmatization
lm = WordNetLemmatizer()

lemmatized = [lm.lemmatize(word) for word in filtered]

print("\nLemmatized Words:")
print(lemmatized)

# 6. TF-IDF
vector = TfidfVectorizer()

tfidf = vector.fit_transform([text])

print("\nTF-IDF Words:")
print(vector.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf.toarray())
```
