1. Locate open Iris dataset from the URL csv_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
2. Perform the label encoding , by considering Species as target variable.


```python
import pandas as pd   
import numpy as np          
import seaborn as sns      
import matplotlib.pyplot as plt  
```

```python
url="https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
```

```python
df=pd.read_csv(url,header=None,names=["sep_len","sep_wid","pet_len","pet_wid","spe"])
df.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
df["spe"]=le.fit_transform(df["spe"])
df.head(10),df.tail(10)
```
